```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
flights=sns.load_dataset('flights')
flights
```

```python
iris=sns.load_dataset('iris')
iris.head()
```

```python
tips=sns.load_dataset('tips')
tips
```

1. line & scatter plot using matplotlib:

```python
for month in flights['month'].unique():
    data = flights[flights['month'] == month]
    plt.plot(data['year'], data['passengers'], label=month, marker='.')
plt.title('Passangers groth over years', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Passengers')

plt.legend()
plt.show()
```

2. line & scatter plot using Seaborn:

```python
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species')
plt.title('Relation between sepal length and sepal width', fontsize=14)
plt.xlabel('sepal length')
plt.ylabel('sepal width')
plt.show()
```

```python

sns.lineplot(data=flights,x='year',y='passengers', hue='month', marker='o', alpha=0.5)
plt.title('Passangers groth over years', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Passengers')
plt.show()
```

3. KDE plots

```python
sns.histplot(data=flights, x='passengers', color='cyan' , kde=True , bins=10)
plt.title('Passangers growth over years', fontsize=14)
plt.show()
```

```python
sns.kdeplot(data=tips, x="total_bill", hue="sex", fill=True, alpha=0.5, palette='rocket' )
plt.title("Total Bill Distribution by Gender (KDE)", fontsize=14)
plt.xlabel("Total Bill ($)")
plt.show()
```

4.  ECDF Plot

```python

sns.ecdfplot(data=flights,x='passengers', hue='month', marker='o', alpha=0.5)
plt.title('Passangers groth over years', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Passengers')
plt.show()
```

```python
sns.ecdfplot(data=tips, x='total_bill', hue='sex', palette='rocket')
plt.title("Total Bill Distribution by Gender (KDE)", fontsize=14)
plt.xlabel("Total Bill ($)")
plt.show()

```

5. Categorical & Comparison Plots

* Bar plot

```python
x=tips['size']
y=tips['tip']
plt.bar(x,y, color='#8357c9')
plt.title('Relation between size and tip', fontsize=14)
plt.xlabel('Size')
plt.ylabel('Tip')
plt.show()
```

```python

sns.barplot(data=tips, x='size' ,y='tip' , hue='sex', palette='dark:#8357c9')
plt.title('Relation between size and tip', fontsize=14)
plt.xlabel('Size')
plt.ylabel('Tip')
plt.show()
```

```python
diamonds=sns.load_dataset("diamonds")
color_count=diamonds['color'].value_counts()
plt.bar(color_count.index, color_count.values,
        color="skyblue",
        align="center",
        edgecolor="black")
plt.title("Distribution of Diamond Color Grades", fontsize=14)
plt.xlabel("Diamond Color Grade (D = Best, J = Worst)", fontsize=12)
plt.ylabel("Number of Diamonds", fontsize=12)
plt.show()
```

* Box plot

```python
sns.boxplot(data=flights, x='month', y='passengers', palette='rocket')
plt.title('Passengers distribution in month', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Passengers')
plt.show()

```

* Violin Plot

```python
sns.violinplot(data=flights, x='month', y='passengers', palette='rocket')
plt.title('Passengers distribution in month', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Passengers')
plt.show()
```
