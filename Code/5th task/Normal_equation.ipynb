Step 2: Splitting the dataset

```python
data=data.sample(frac=1, random_state=1).reset_index(drop=True)
n=len(data)
training_set = data[:int(n*0.7)]
validation_set = data[int(n*0.7) : int(n*0.85)]
test_set = data[int(n*0.85):]
```

```python

y_train = training_set['Median_House_Value']
X_train = training_set.drop(columns=['Median_House_Value'])
y_val = validation_set['Median_House_Value']
X_val = validation_set.drop(columns=['Median_House_Value'])
y_test = test_set['Median_House_Value']
X_test = test_set.drop(columns=['Median_House_Value'])

```

Linear Regression — Normal Equation

Step 1: add the bias column

```python

X_train_b = np.concatenate((np.ones((X_train.shape[0],1)), X_train.values),axis=1)
X_val_b = np.concatenate((np.ones((X_val.shape[0],1)), X_val.values),axis=1)
X_tes_b = np.concatenate((np.ones((X_test.shape[0],1)), X_test.values),axis=1)
```

Step 2: implement the normal equation

```python
w = np.linalg.inv(X_train_b.T @ X_train_b) @ X_train_b.T @ y_train.values
```

Step 3: make a predection

```python
y_val_predicted = X_val_b @ w
```

Loss function:

```python
def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)
def mae_loss(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))
def huber_loss(y_true, y_pred, delta=1.0):
    error = y_true - y_pred
    is_small = np.abs(error) <= delta
    squared_loss= 0.5 * error**2
    linear_loss= delta * (np.abs(error)- 0.5 * delta)
    return np.mean(np.where(is_small, squared_loss, linear_loss))


```

Evaluation of the first model:

```python
print("MSE= ", mse_loss(y_val, y_val_predicted))
print("MAE= ", mae_loss(y_val, y_val_predicted))
```
