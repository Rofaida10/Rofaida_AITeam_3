```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import euler
```

```python
x=list(range(-100,100))
y=1 / (1 + np.exp(np.negative(x)))
plt.plot(x,y,linewidth=2,marker='.',markerfacecolor='yellow',markersize=15)
plt.xlabel('x',fontsize=15)
plt.ylabel('σ(x)',fontsize=15)
plt.title('Sigmoid activation function',fontsize=20)
plt.show()
```
